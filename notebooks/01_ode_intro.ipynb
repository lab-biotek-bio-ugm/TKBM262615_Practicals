```python
# Setup — Colab already ships numpy, scipy, and matplotlib, so no pip install is needed.
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
```

## Key Concepts

- **Static vs. dynamic views:** a pathway diagram only shows static relationships (who interacts with whom); dynamic models add *time* and *magnitude* (how fast, how large).
- **State variables:** quantities that change with time and describe the "state" of the system — e.g. protein concentration $X(t)$, population size $N(t)$, membrane potential $V(t)$.
- **Parameters:** fixed (constant) quantities that determine behavior — e.g. reaction rate constant $k$, carrying capacity $K$.
- **Rate of change / derivative:** $\frac{dx}{dt}$ — positive means increasing, negative means decreasing, zero means *steady state*.
- **Ordinary differential equation (ODE):** an equation relating a function to its derivative(s). First-order ODE: $\frac{dx}{dt} = f(x)$.

## Key Equations

- Exponential growth:
$$ \frac{dN}{dt} = rN, \qquad N(t) = N_0 e^{rt} $$
  - $r > 0$: unbounded growth; $r < 0$: towards extinction.

- Logistic growth (carrying capacity $K$):
$$ \frac{dN}{dt} = rN\left(1 - \frac{N}{K}\right) $$
  - *steady states*: $N = 0$ (unstable) and $N = K$ (stable).

- Birth–degradation (birth rate $k_{in}$, degradation $k_{deg}$):
$$ \frac{dx}{dt} = k_{in} - k_{deg}\, x \quad\Rightarrow\quad x_{ss} = \frac{k_{in}}{k_{deg}} $$


```python
# Model 1: exponential growth  dN/dt = r N
def exp_model(t, N, r):
    return r * N

# Model 2: logistic growth  dN/dt = r N (1 - N/K)
def logistic_model(t, N, r, K):
    return r * N * (1 - N / K)

# Model 3: birth-death  dx/dt = k_in - k_deg x
def birth_death(t, x, k_in, k_deg):
    return k_in - k_deg * x
```

```python
r, K = 0.5, 100.0          # logistic parameters
t_eval = np.linspace(0, 10, 400)

sol_exp = solve_ivp(exp_model, (0, 10), [1.0], t_eval=t_eval, args=(r,))
sol_log = solve_ivp(logistic_model, (0, 10), [1.0], t_eval=t_eval, args=(r, K))
sol_bd = solve_ivp(birth_death, (0, 10), [0.0], t_eval=t_eval, args=(2.0, 0.5))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(13, 4))
axes[0].plot(sol_exp.t, sol_exp.y[0])
axes[0].set_title("Exponential $dN/dt = rN$")
axes[1].plot(sol_log.t, sol_log.y[0])
axes[1].set_title(f"Logistic (K = {K})")
axes[2].plot(sol_bd.t, sol_bd.y[0])
axes[2].set_title("Birth-death")
for ax in axes:
    ax.set_xlabel("time t")
    ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Logistic steady state   : N -> {K}")
print(f"Birth-death steady state: x_ss = k_in/k_deg = {2.0/0.5}")
```
